### 📖 Concept: The Series
A Series is a one-dimensional array with labeled indices. It's the building block of a DataFrame.

```python
# [DEMO] Creating a Series
# [4, 7, -5, 3] a Python list — ordered group of values, in square brackets.
# index=[...] give each value a name/label instead of just number 0,1,2,3.
obj = pd.Series([4, 7, -5, 3], index=["d", "b", "a", "c"])
obj  # last line in cell auto-display, no print() needed in Jupyter
```

### 📖 Concept: The DataFrame
A DataFrame is a 2-dimensional labeled data structure. Think of it as a dictionary of Series objects (each column is a Series).

```python
# [DEMO] Creating a DataFrame from a dictionary
# {} curly braces = Python dictionary: "key": [values...] pairs.
# Each key become a column name, its list become that column's values.
data = {
    "state": ["Ohio", "Ohio", "Ohio", "Nevada", "Nevada", "Nevada"],
    "year": [2000, 2001, 2002, 2001, 2002, 2003],
    "pop": [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]
}

frame = pd.DataFrame(data)  # turn dictionary into table
frame
```

```python
# .head(2) show first 2 rows only. Dot after frame call a "method" (action) on it.
frame.head(2)
```

### 💡 Concept Demo: Inspecting & Modifying
1. `head()` and `tail()`
2. Accessing columns (dictionary syntax vs dot syntax)
3. Adding a new column

```python
# [DEMO] Inspecting data
print("First 5 rows:")
display(frame.head())  # display() like print but render table nicely (Jupyter only)

# [DEMO] Accessing a column
# frame['pop'] grab one column by name, in quotes (string), inside square brackets.
print("Population column:")
display(frame['pop'])
```

```python
# [DEMO] Adding a new column with a scalar value
# frame['debt'] = 16.5 create new column named "debt", same value every row.
frame['debt'] = 16.5
frame
```

```python
# [DEMO] Adding a calculated column
# == compare each row's state to 'Ohio' — True if match, False if not.
# Result (True/False per row) become new column "eastern".
frame['eastern'] = frame.state == 'Ohio'
frame
```

### 📖 Concept: Detecting Missing Data
Real-world data often has missing values (shown as `NaN`). Use `isna()` and `notna()` to detect them.

```python
# [DEMO] Detecting missing values
# np.nan = "Not a Number", pandas/numpy's way of marking missing/empty data.
data_with_nan = pd.Series([1, np.nan, 3.5, np.nan, 7])

print("Original data:")
display(data_with_nan)

print("\nWhich values are missing?")
display(data_with_nan.isna())  # isna() → True where value is NaN

print("\nWhich values are NOT missing?")
display(data_with_nan.notna())  # notna() → opposite of isna()
```

## Part 2: Indexing and Selection (60 min)

### Learning Objectives
By the end of this section, you will be able to:
- **Differentiate** between label-based (`loc`) and position-based (`iloc`) indexing.
- **Apply** boolean filtering to select specific rows.
- **Drop** rows or columns from a DataFrame.
- **Modify** specific data points using selection.

```python
# Setup for Demo
# np.arange(16) make list of numbers 0-15. .reshape((4,4)) fold it into 4 rows x 4 cols grid.
data = pd.DataFrame(np.arange(16).reshape((4, 4)),
                    index=["Ohio", "Colorado", "Utah", "New York"],
                    columns=["one", "two", "three", "four"])
data
```

```python
# .loc["Colorado"] grab row by its label (name), not position number.
data.loc["Colorado"]
```

### 💡 Concept Demo: Loc vs Iloc
- **.loc[row_label, col_label]**: "Address" based.
- **.iloc[row_pos, col_pos]**: "GPS" based (integers).

```python
# [DEMO] Select 'Colorado' row using loc
print("--- Row by Label ---")
display(data.loc["Colorado"])

# [DEMO] Select 'Colorado' row using iloc (it's at index 1)
# iloc use plain position number (0,1,2,...) like list index, ignore row label.
print("\n--- Row by Position ---")
display(data.iloc[1])
```

```python
# [DEMO] Boolean Filtering
# Select rows where column 'three' is greater than 5
# data["three"] > 5 make True/False per row; data[...] keep only True rows.
data[data["three"] > 5]
```

### 📖 Concept: Dropping Rows and Columns
Use the `drop()` method to remove rows or columns. By default, it drops rows. Use `axis=1` or `axis="columns"` to drop columns.

```python
# [DEMO] Dropping rows and columns
print("Drop rows 'Colorado' and 'Ohio':")
display(data.drop(["Colorado", "Ohio"]))  # drop() default remove rows by label

print("\nDrop column 'two':")
display(data.drop("two", axis=1))  # axis=1 tell pandas "two" a column, not row
```

## Part 3: Function Application, Sorting, & Ranking (60 min)

### Learning Objectives
By the end of this section, you will be able to:
- **Sort** data by index or values.
- **Rank** data to handle ties.
- **Apply** custom functions to transform data.

```python
# Setup Data
frame = pd.DataFrame({
    "b": [4, 7, -3, 2], 
    "a": [0, 1, 0, 1]
}, index=['Row1', 'Row2', 'Row3', 'Row4'])
frame
```

### 💡 Concept Demo: Sorting & Ranking
1. `sort_index()`
2. `sort_values(by=...)`
3. `rank()`

```python
# [DEMO] Sort by values in column 'b'
frame.sort_values(by="b")
```

```python
# [DEMO] Applying a function
# def = "define a function": a reusable block of code you name and call later.
# val is a placeholder name for whatever value get passed in when function run.
def label_size(val):
    return "Big" if val > 2 else "Small"  # one-line if/else: shorthand for if-else

# .apply(label_size) run label_size() on every value in column 'b', one by one.
frame['b_size'] = frame['b'].apply(label_size)
frame
```
